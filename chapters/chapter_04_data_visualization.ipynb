```python
# Essential imports for visualization
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Set style for better-looking plots
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

# Load our sample data
house_data = pd.read_csv('../datasets/sample_house_data.csv')
customer_data = pd.read_csv('../datasets/customer_data.csv')

print("Data loaded successfully!")
print(f"House data shape: {house_data.shape}")
print(f"Customer data shape: {customer_data.shape}")
```

```python
# Basic Matplotlib Plots
print("--- Basic Matplotlib Examples ---")

# Create a figure with subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Basic Plot Types', fontsize=16)

# 1. Line plot
x = np.linspace(0, 10, 100)
y = np.sin(x)
axes[0, 0].plot(x, y, linewidth=2)
axes[0, 0].set_title('Line Plot')
axes[0, 0].set_xlabel('X values')
axes[0, 0].set_ylabel('Y values')
axes[0, 0].grid(True, alpha=0.3)

# 2. Scatter plot
axes[0, 1].scatter(house_data['Size'], house_data['Price'], alpha=0.7)
axes[0, 1].set_title('House Size vs Price')
axes[0, 1].set_xlabel('Size (sq ft)')
axes[0, 1].set_ylabel('Price ($)')

# 3. Histogram
axes[1, 0].hist(house_data['Age'], bins=8, alpha=0.7, edgecolor='black')
axes[1, 0].set_title('Distribution of House Ages')
axes[1, 0].set_xlabel('Age (years)')
axes[1, 0].set_ylabel('Frequency')

# 4. Bar plot
bedroom_counts = house_data['Bedrooms'].value_counts().sort_index()
axes[1, 1].bar(bedroom_counts.index, bedroom_counts.values, alpha=0.7)
axes[1, 1].set_title('Houses by Number of Bedrooms')
axes[1, 1].set_xlabel('Number of Bedrooms')
axes[1, 1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
# Seaborn for Beautiful Statistical Plots
print("--- Seaborn Statistical Plots ---")

# Create a figure with seaborn plots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Seaborn Statistical Visualizations', fontsize=16)

# 1. Distribution plot
sns.histplot(data=house_data, x='Price', kde=True, ax=axes[0, 0])
axes[0, 0].set_title('House Price Distribution')

# 2. Box plot
# Create age categories for box plot
house_data['Age_Cat'] = pd.cut(house_data['Age'], bins=[0, 5, 15, 25], labels=['New', 'Medium', 'Old'])
sns.boxplot(data=house_data, x='Age_Cat', y='Price', ax=axes[0, 1])
axes[0, 1].set_title('Price by Age Category')

# 3. Correlation heatmap
numeric_cols = house_data.select_dtypes(include=[np.number]).columns
correlation_matrix = house_data[numeric_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, ax=axes[1, 0])
axes[1, 0].set_title('Correlation Matrix')

# 4. Pair plot (we'll show just a subset)
# For the pair plot, we'll create a separate figure since it works better standalone
subset_data = house_data[['Size', 'Bedrooms', 'Age', 'Price']].sample(10)  # Small subset for clarity
axes[1, 1].scatter(subset_data['Size'], subset_data['Price'], s=subset_data['Bedrooms']*20, 
                   c=subset_data['Age'], cmap='viridis', alpha=0.7)
axes[1, 1].set_xlabel('Size')
axes[1, 1].set_ylabel('Price')
axes[1, 1].set_title('Size vs Price (Bubble: Bedrooms, Color: Age)')

plt.tight_layout()
plt.show()
```

```python
# Exploratory Data Analysis (EDA) Example
print("--- Comprehensive EDA Example ---")

def perform_eda(data, title):
    """Perform basic EDA on a dataset"""
    print(f"\n=== {title} ===")
    print(f"Shape: {data.shape}")
    print(f"\nFirst few rows:")
    print(data.head())
    
    print(f"\nBasic Statistics:")
    print(data.describe())
    
    print(f"\nMissing Values:")
    print(data.isnull().sum())
    
    # Visualization
    numeric_columns = data.select_dtypes(include=[np.number]).columns
    if len(numeric_columns) > 1:
        plt.figure(figsize=(12, 8))
        
        # Distribution plots for numeric columns
        for i, col in enumerate(numeric_columns[:4]):  # Limit to 4 columns
            plt.subplot(2, 2, i+1)
            plt.hist(data[col], bins=10, alpha=0.7, edgecolor='black')
            plt.title(f'Distribution of {col}')
            plt.xlabel(col)
            plt.ylabel('Frequency')
        
        plt.tight_layout()
        plt.show()

# Perform EDA on our datasets
perform_eda(house_data, "House Dataset")
perform_eda(customer_data, "Customer Dataset")
```

```python
# Advanced Visualization: Custom Plots
print("--- Advanced Custom Visualizations ---")

# Create a professional-looking dashboard
fig = plt.figure(figsize=(16, 12))
gs = fig.add_gridspec(3, 3, hspace=0.3, wspace=0.3)

# Main scatter plot
ax1 = fig.add_subplot(gs[0:2, 0:2])
scatter = ax1.scatter(house_data['Size'], house_data['Price'], 
                     c=house_data['Age'], s=house_data['Bedrooms']*30, 
                     alpha=0.6, cmap='viridis', edgecolors='black', linewidth=0.5)
ax1.set_xlabel('House Size (sq ft)', fontsize=12)
ax1.set_ylabel('Price ($)', fontsize=12)
ax1.set_title('House Price Analysis Dashboard', fontsize=14, fontweight='bold')
ax1.grid(True, alpha=0.3)

# Add colorbar for age
cbar = plt.colorbar(scatter, ax=ax1)
cbar.set_label('House Age (years)', fontsize=11)

# Histogram of prices
ax2 = fig.add_subplot(gs[0, 2])
ax2.hist(house_data['Price'], bins=8, alpha=0.7, color='skyblue', edgecolor='black')
ax2.set_title('Price Distribution')
ax2.set_xlabel('Price ($)')
ax2.set_ylabel('Count')

# Box plot of price by bedrooms
ax3 = fig.add_subplot(gs[1, 2])
bedroom_data = [house_data[house_data['Bedrooms'] == br]['Price'].values for br in sorted(house_data['Bedrooms'].unique())]
ax3.boxplot(bedroom_data, labels=sorted(house_data['Bedrooms'].unique()))
ax3.set_title('Price by Bedrooms')
ax3.set_xlabel('Bedrooms')
ax3.set_ylabel('Price ($)')

# Summary statistics table
ax4 = fig.add_subplot(gs[2, :])
ax4.axis('off')
summary_stats = house_data[['Size', 'Bedrooms', 'Bathrooms', 'Age', 'Price']].describe().round(2)
table = ax4.table(cellText=summary_stats.values, 
                 rowLabels=summary_stats.index,
                 colLabels=summary_stats.columns,
                 cellLoc='center',
                 loc='center')
table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1.2, 1.5)
ax4.set_title('Summary Statistics', fontsize=12, fontweight='bold', pad=20)

plt.show()

print("\n📊 Dashboard created! This shows:")
print("- Main plot: Size vs Price (color=age, size=bedrooms)")
print("- Price distribution histogram")
print("- Price by number of bedrooms")
print("- Summary statistics table")
```

## Visualization Best Practices

### Choosing the Right Plot Type:
- **Scatter Plot**: Relationship between two continuous variables
- **Line Plot**: Trends over time or ordered data
- **Histogram**: Distribution of a single variable
- **Box Plot**: Distribution comparison across categories
- **Heatmap**: Correlation matrices or 2D data
- **Bar Plot**: Categorical data comparison

### Design Principles:
1. **Clear titles and labels**: Always label your axes and add titles
2. **Appropriate colors**: Use color purposefully, not just decoration
3. **Clean layout**: Remove unnecessary elements (chart junk)
4. **Consistent scale**: Make sure scales are appropriate for your data
5. **Readable text**: Use appropriate font sizes

### Common Mistakes to Avoid:
- Too many colors or patterns
- Unclear or missing labels
- Inappropriate plot types
- Cluttered layouts
- Misleading scales
