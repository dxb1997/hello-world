# Chapter 8: Model Evaluation and Validation

## Learning Objectives
- Master cross-validation
- Learn hyperparameter tuning
- Understand evaluation metrics

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import cross_val_score, GridSearchCV
from sklearn.ensemble import RandomForestClassifier

# Load data
customer_data = pd.read_csv('../datasets/customer_data.csv')
customer_data['Gender_Numeric'] = customer_data['Gender'].map({'Male': 0, 'Female': 1})

X = customer_data[['Age', 'Income', 'SpendingScore', 'Gender_Numeric']]
y = customer_data['Purchased']

# Cross-validation
model = RandomForestClassifier(random_state=42)
cv_scores = cross_val_score(model, X, y, cv=5)

print(f"CV scores: {cv_scores}")
print(f"Mean: {cv_scores.mean():.3f} (+/- {cv_scores.std() * 2:.3f})")

# Hyperparameter tuning
param_grid = {'n_estimators': [10, 50, 100], 'max_depth': [3, 5, 10]}
grid_search = GridSearchCV(model, param_grid, cv=3)
grid_search.fit(X, y)

print(f"Best params: {grid_search.best_params_}")
print(f"Best score: {grid_search.best_score_:.3f}")
```
