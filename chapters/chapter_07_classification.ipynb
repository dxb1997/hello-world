## 🎯 Compare Multiple Algorithms

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Load data
customer_data = pd.read_csv('../datasets/customer_data.csv')
customer_data['Gender_Numeric'] = customer_data['Gender'].map({'Male': 0, 'Female': 1})

X = customer_data[['Age', 'Income', 'SpendingScore', 'Gender_Numeric']]
y = customer_data['Purchased']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Compare algorithms
models = {
    'Logistic Regression': LogisticRegression(random_state=42),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42)
}

results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    results[name] = accuracy
    print(f"{name}: {accuracy:.3f}")

# Visualize
plt.figure(figsize=(10, 6))
plt.bar(results.keys(), results.values())
plt.title('Classification Model Comparison')
plt.ylabel('Accuracy')
plt.show()
```
