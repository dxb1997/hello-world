```python
# Introduction to Pandas
import pandas as pd
import numpy as np

# Creating DataFrames - the heart of pandas
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'London', 'Tokyo', 'Paris'],
    'Salary': [50000, 60000, 55000, 58000]
}

df = pd.DataFrame(data)
print("Sample DataFrame:")
print(df)
print("\nDataFrame Info:")
print(f"Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
print(f"Data types:\n{df.dtypes}")
```

```python
# Loading Real Data - Using our sample datasets
print("Loading sample datasets...")

# Load house data
house_data = pd.read_csv('../datasets/sample_house_data.csv')
print("House Data:")
print(house_data.head())
print(f"\nShape: {house_data.shape}")

# Basic exploration
print("\nBasic Statistics:")
print(house_data.describe())

# Check for missing values
print("\nMissing Values:")
print(house_data.isnull().sum())
```

```python
# Data Selection and Filtering
print("--- Data Selection Examples ---")

# Select specific columns
print("Houses with size and price only:")
size_price = house_data[['Size', 'Price']]
print(size_price.head())

# Filter rows based on conditions
print("\nLarge houses (>2000 sq ft):")
large_houses = house_data[house_data['Size'] > 2000]
print(large_houses)

# Multiple conditions
print("\nLarge, new houses (>2000 sq ft and <10 years old):")
large_new = house_data[(house_data['Size'] > 2000) & (house_data['Age'] < 10)]
print(large_new)
```

```python
# Data Transformation and Feature Engineering
print("--- Data Transformation Examples ---")

# Create new features
house_data_copy = house_data.copy()

# Price per square foot
house_data_copy['Price_per_sqft'] = house_data_copy['Price'] / house_data_copy['Size']

# Categorize houses by age
def categorize_age(age):
    if age < 5:
        return 'New'
    elif age < 15:
        return 'Medium'
    else:
        return 'Old'

house_data_copy['Age_Category'] = house_data_copy['Age'].apply(categorize_age)

# Total rooms
house_data_copy['Total_Rooms'] = house_data_copy['Bedrooms'] + house_data_copy['Bathrooms']

print("Enhanced dataset with new features:")
print(house_data_copy[['Size', 'Price', 'Price_per_sqft', 'Age_Category', 'Total_Rooms']].head())
```

```python
# Handling Missing Data (Simulated)
print("--- Handling Missing Data ---")

# Create a copy with some missing values for demonstration
data_with_missing = house_data.copy()
# Simulate missing values
data_with_missing.loc[1, 'Bedrooms'] = np.nan
data_with_missing.loc[3, 'Age'] = np.nan
data_with_missing.loc[5, 'Price'] = np.nan

print("Data with missing values:")
print(data_with_missing.isnull().sum())

# Strategy 1: Fill with mean
print("\nFilling missing values with mean:")
filled_mean = data_with_missing.copy()
for column in ['Bedrooms', 'Age', 'Price']:
    if filled_mean[column].isnull().any():
        mean_value = filled_mean[column].mean()
        filled_mean[column] = filled_mean[column].fillna(mean_value)
        print(f"{column}: filled with {mean_value:.1f}")

# Strategy 2: Forward fill
print("\nForward fill example:")
filled_forward = data_with_missing.fillna(method='ffill')
print("Missing values after forward fill:", filled_forward.isnull().sum().sum())

# Strategy 3: Drop rows with missing values
print("\nDropping rows with missing values:")
dropped_na = data_with_missing.dropna()
print(f"Original shape: {data_with_missing.shape}")
print(f"After dropping: {dropped_na.shape}")
```

## Key Pandas Operations for ML

### Essential DataFrame Operations:
- **`.head()`**: View first few rows
- **`.describe()`**: Statistical summary
- **`.info()`**: DataFrame information
- **`.shape`**: Dimensions of data
- **`.columns`**: Column names

### Data Selection:
- **`df['column']`**: Select single column
- **`df[['col1', 'col2']]`**: Select multiple columns
- **`df[condition]`**: Filter rows

### Data Cleaning:
- **`.isnull()`**: Check for missing values
- **`.fillna()`**: Fill missing values
- **`.dropna()`**: Drop missing values
- **`.duplicated()`**: Find duplicates
