# Chapter 9: Unsupervised Learning

## Learning Objectives
- Master K-Means clustering
- Learn PCA for dimensionality reduction
- Apply clustering to real data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Load and prepare data
customer_data = pd.read_csv('../datasets/customer_data.csv')
X = customer_data[['Age', 'Income', 'SpendingScore']]

# Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# K-Means clustering
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(X_scaled)

# Visualize clusters
plt.figure(figsize=(10, 6))
plt.scatter(X['Income'], X['SpendingScore'], c=clusters, cmap='viridis')
plt.xlabel('Income')
plt.ylabel('Spending Score')
plt.title('Customer Segments')
plt.colorbar()
plt.show()

print(f"Cluster sizes: {np.bincount(clusters)}")
```

```python
# Principal Component Analysis
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print(f"Explained variance: {pca.explained_variance_ratio_}")
print(f"Total: {pca.explained_variance_ratio_.sum():.3f}")

# Visualize PCA
plt.figure(figsize=(10, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=clusters, cmap='viridis')
plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.3f})')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.3f})')
plt.title('PCA with Clusters')
plt.colorbar()
plt.show()
```
