```python
# Essential imports for supervised learning
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error, accuracy_score
from sklearn.preprocessing import StandardScaler

print("📚 Chapter 5: Introduction to Supervised Learning")
print("✅ All libraries imported successfully!")
```

## What is Supervised Learning?

**Supervised Learning** is like learning with a teacher. We provide the algorithm with:
- **Input data (features)**: What we know about each example
- **Output data (labels/targets)**: What we want to predict

### Two Main Types:
1. **Regression**: Predicting continuous values (prices, temperatures, scores)
2. **Classification**: Predicting categories (spam/not spam, pass/fail, cat/dog)

### The Process:
1. **Training**: Algorithm learns patterns from labeled data
2. **Testing**: We evaluate how well it predicts on new, unseen data
3. **Prediction**: Use the trained model on real-world data

```python
# Train-Test Split: The Foundation of Supervised Learning
print("--- Train-Test Split Example ---")

# Load our house data
house_data = pd.read_csv('../datasets/sample_house_data.csv')
print(f"Dataset shape: {house_data.shape}")
print(house_data.head())

# Prepare features (X) and target (y)
X = house_data[['Size', 'Bedrooms', 'Bathrooms', 'Age']]  # Features
y = house_data['Price']  # Target

print(f"\nFeatures shape: {X.shape}")
print(f"Target shape: {y.shape}")

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print(f"\nTraining set: {X_train.shape[0]} samples")
print(f"Testing set: {X_test.shape[0]} samples")
print(f"Split ratio: {X_train.shape[0]/(X_train.shape[0]+X_test.shape[0]):.1%} train, {X_test.shape[0]/(X_train.shape[0]+X_test.shape[0]):.1%} test")
```

```python
# Building Your First Supervised Model
print("--- Building a Regression Model ---")

# Create and train the model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions
y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)

# Evaluate the model
train_mse = mean_squared_error(y_train, y_pred_train)
test_mse = mean_squared_error(y_test, y_pred_test)

print(f"Training MSE: {train_mse:.2f}")
print(f"Testing MSE: {test_mse:.2f}")

# Feature importance (coefficients)
feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': model.coef_
})
print("\nFeature Importance:")
print(feature_importance)

# Visualize predictions vs actual
plt.figure(figsize=(10, 6))
plt.scatter(y_test, y_pred_test, alpha=0.7)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Actual vs Predicted House Prices')
plt.show()

print("\n🎯 The closer points are to the red line, the better our predictions!")
```

```python
# Classification Example
print("--- Building a Classification Model ---")

# Load customer data for classification
customer_data = pd.read_csv('../datasets/customer_data.csv')
print(f"Customer data shape: {customer_data.shape}")

# Prepare features and target for classification
# Convert Gender to numeric
customer_data['Gender_Numeric'] = customer_data['Gender'].map({'Male': 0, 'Female': 1})

X_class = customer_data[['Age', 'Income', 'SpendingScore', 'Gender_Numeric']]
y_class = customer_data['Purchased']

# Split the data
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
    X_class, y_class, test_size=0.3, random_state=42
)

# Scale the features (important for many algorithms)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_class)
X_test_scaled = scaler.transform(X_test_class)

# Train classification model
clf_model = LogisticRegression(random_state=42)
clf_model.fit(X_train_scaled, y_train_class)

# Make predictions
y_pred_class = clf_model.predict(X_test_scaled)
y_pred_proba = clf_model.predict_proba(X_test_scaled)[:, 1]  # Probability of purchase

# Evaluate
accuracy = accuracy_score(y_test_class, y_pred_class)
print(f"\nClassification Accuracy: {accuracy:.2%}")

# Show some predictions
results_df = pd.DataFrame({
    'Actual': y_test_class.values,
    'Predicted': y_pred_class,
    'Probability': y_pred_proba
})
print("\nSample Predictions:")
print(results_df.head(10))
```

## Key Concepts Summary

### Supervised Learning Workflow:
1. **Data Preparation**: Clean and prepare your dataset
2. **Feature Selection**: Choose relevant input variables
3. **Train-Test Split**: Separate data for training and evaluation
4. **Model Training**: Algorithm learns from training data
5. **Evaluation**: Test model performance on unseen data
6. **Prediction**: Use model on new data

### Important Terms:
- **Features (X)**: Input variables used for prediction
- **Target (y)**: What we want to predict
- **Training Set**: Data used to train the model
- **Test Set**: Data used to evaluate model performance
- **Overfitting**: Model memorizes training data but fails on new data
- **Underfitting**: Model is too simple to capture patterns
